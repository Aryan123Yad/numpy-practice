```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

matplotlib basis


```python
x = np.linspace(0,5,11)
x
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title('growth year')
plt.xlabel('time')
plt.ylabel('adani')
```

subplot in matplotlib
ek sathmultiple graph create karna

```python
plt.subplot(2,2,1)
plt.plot(x,y,'r')
plt.subplot(2,2,2)
plt.plot(y,x,'b')
plt.subplot(2,2,3)
plt.plot(x,y,'g')
plt.subplot(2,2,4)
plt.plot(x**4,x**4)
```

object orinted way
matplotlib allows the aspect ratio, dpi and figure size to be specified when the figureobject is creatd.you can ues thr figsize and dpi keyword argument.

figsize is a tuple of the width and hight of the figure in inches dpi is the dots per inch (pixel in inch.

```python
fig = plt.figure(figsize=(8,4),dpi=100)

fig =plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y)
axes.set_xlabel('x',fontsize=15)
axes.set_ylabel('y',fontsize=15)
axes.set_title('title',fontsize=20)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.2,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.6,0.4,0.2,0.3])
axis2.plot(x,y)

#issi me change karo ge sab upper niche hoga kahe aur badalne ki jarurt nahi hai
```
