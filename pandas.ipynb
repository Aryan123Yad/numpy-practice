merging 2 dataframes


```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
    'emp_id':[1,2,3,4,5],
    'name':['john','rahul','shubham','rupesh',"shivam"],
    'age':[28,34,54,55,60],
})

#dataframe2 :salary information
salaries =pd.DataFrame({
    'emp_id':[1,2,3,4,5],
    'salary':['50000','30000','45000','34000','40000']
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on='emp_id',how ='inner')
```

```python
pd.merge(employees,salaries,on='emp_id',how ='outer')
```

```python
pd.merge(employees,salaries,on='emp_id',how ='right')
```

groupby

```python
data = {
    'categary':['A','B','C','A','B','C'],
    'store':['store1','store1','store2','store2','store1','store2'],
    'sales':[100,200,300,400,500,600],
    "Quantity":[10,20,30,40,50,60],
    'date':pd.date_range('2023-01-01',periods=6)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
#group by category and calculate the sum of sales
Cat= df.groupby('categary')
Cat
for i, v in Cat:
  print(i)
  print(v)
```

```python
df.groupby('categary')['sales'].sum()
```

```python
#group by store and calculate the sum of sales
df.groupby('store')['sales'].sum()
```

```python
#group by multtiple colums
#group by category and store
cat = df.groupby(['categary','store'])['sales'].sum()
cat
```
